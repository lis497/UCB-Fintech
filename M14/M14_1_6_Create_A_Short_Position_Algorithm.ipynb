## Create a DMAC Trading Signal

### Step 1: Read the `vnq.csv` file from the Resources folder into a Pandas DataFrame by. Set the “Date” column as the index. Review the DataFrame.

```python
# Read the vnq.csv file from the Resouces folder into a Pandas DataFrame
# Set the `Date` column as the DateTimeIndex
vnq_df = pd.read_csv(
    Path("14.1/02-Create_a_Short_Position_Algorithm/Resources/vnq.csv"), 
    index_col="Date",
    parse_dates=True 
    
)

# Review the DataFrame
vnq_df.head()
```

## Step 2: By filtering the DataFrame, create a second DataFame, named `signals_df`, that filters out only the “Date” index and the “Close” column. Review the new DataFrame, and then use hvPlot to visualize the pricing information.

```python
signals_df = vnq_df.loc[:, ['Close']]
signals_df.head()
```

```python
signals_df.plot()
```

### Step 3: Set the `short_window` and `long_window` variables to 50 and 100, respectively.

```python
# Set the short_window (50) and long window (100) variables
short_window = 50
long_window = 100
```

### Step 4: Using the “Close” columns as well as the `rolling` and `mean` functions, create columns named “SMA50” and “SMA100” that contain the 50-day SMA and a 100-day SMA simple moving averages, respectively. Review the DataFrame.

```python
# Generate the short and long moving averages (50 and 100 days, respectively)
signals_df['SMA50'] = signals_df['Close'].rolling(window=short_window).mean()
signals_df['SMA100'] = signals_df['Close'].rolling(window=long_window).mean()

signals_df.tail()
```

### Step 5: Create a new column, named “Signal”, in the DataFrame, and set all its values to 0.0.

```python
# Initialize the new Signal column to hold the trading signal
signals_df['Signal'] = 0.0
signals_df.head()
```

### Step 6: Using the NumPy `where` function, set the “Signal” column to 1.0 when the “SMA50” value is less than the “SMA100” value. Otherwise, set the “Signal” column to 0.0.

```python
# Generate the trading signal 0 or 1,
# where 1 is the short-window (SMA50) is less than the long-window (SMA100)
signals_df.iloc[short_window:,3:] = np.where(
    signals_df['SMA50'][short_window:] < signals_df['SMA100'][short_window:], 1.0, 0.0

)

signals_df.iloc[95:105,:]
```

### Step 7: Use the Pandas `diff` function on the “Signal” column to assign values of 1 or &minus;1 to the “Exit/Entry” column. These values will indicate the trade entry and exit points in time. Review the DataFrame.

```python
# Calculate the points in time at which a position should be taken, 1 or -1
signals_df['Entry/Exit'] = signals_df['Signal'].diff()

# Review the DataFrame
signals_df.loc["2007-10-20":"2007-10-30",:]
```

```python
x = signals_df.index

plt.figure(figsize=(16,7))
plt.title("VNQ - Short-Position Dual Moving Average Trading Algorithm")
plt.plot(x, signals_df['Close'], label='Close')
plt.plot(x, signals_df['SMA50'], label='SMA50')
plt.plot(x, signals_df['SMA100'], label='SMA100')
plt.scatter(signals_df[signals_df['Signal']==1.0].index, signals_df[signals_df['Signal']==1.0]['SMA50'], color='red',s=20)
plt.scatter(signals_df[signals_df['Entry/Exit']==1.0].index, signals_df[signals_df['Entry/Exit']==1.0]['SMA50'], color='black',s=100)
plt.scatter(signals_df[signals_df['Entry/Exit']== -1.0].index, signals_df[signals_df['Entry/Exit']== -1.0]['SMA50'], color='blue',s=100)


#plt.legend()
plt.show()
```
