## Step 4: Import the two Python libraries that you need to access your environment file and API key: `os` and `load_dotenv` from `dotenv`.

```python
# Import the required libraries and dependencies

# Import the os library
import os

# Import the load_dotenv module from the dotenv library
from dotenv import load_dotenv
```

## Step 5: In the `Nasdaq_API_Key` folder that holds this Jupyter notebook, set up an environment file (`.env`) that holds your Nasdaq API key. Make the variable name NASDAQ_API_KEY. Be sure to place the key’s value in quotation marks.

## Step 6: Use the `load_dotenv` function to load the environment variables from the environment file into your notebook.

```python
# Load .env environment variables into the notebook
load_dotenv()
```

## Step 7: Use the Python `os.getenv` function to access your `NASDAQ_API_KEY` from the loaded environment file. Set it equal to a variable of the same name.

```python
# Use the Pandas os module and getenv function to retrieve the Nasdaq API key from the environment variable file.
# Store the Nasdaq API key in a variable of the same name.
finnhub_api = os.getenv('FINNHUB_API')
```

## Step 8: Use the `type` function to confirm that your key is available for use.

```python
# Using the type funcion, confirm that the Nasdaq API key is available for use in the notebook. 
type(finnhub_api)
```
