```python
df_early = pd.merge(coinbase_sliced['Close'].loc['2018-01-16'],bitstamp_sliced['Close'].loc['2018-01-16'],on='Timestamp')
df_early.head()
```

```python
df_early['absolute'] = abs(coinbase_sliced['Close'].loc['2018-01-16'] - bitstamp_sliced['Close'].loc['2018-01-16'])
df_early['rate'] = 0.00000
df_early.loc[df_early['Close_x'] > df_early['Close_y'],'rate']= round(df_early['absolute']/df_early['Close_y'],6)
df_early.loc[df_early['Close_x'] < df_early['Close_y'],'rate']= round(df_early['absolute']/df_early['Close_x'],6)
df_early.head()
```

```python
df_early2 = df_early[df_early['rate'] > 0.01]
df_early2['cumsum'] = df_early2['absolute'].cumsum()
df_early2.head()
```

#### 2. For each of the three dates, calculate the spread returns. To do so, divide the instances that have a positive arbitrage spread (that is, a spread greater than zero) by the price of Bitcoin from the exchange you’re buying on (that is, the lower-priced exchange). Review the resulting DataFrame.

```python
# For the date early in the dataset, calculate the spread returns by dividing the instances when the arbitrage spread is positive (> 0) 
# by the price of Bitcoin from the exchange you are buying on (the lower-priced exchange).
arbitrage_early['spread'] = arbitrage_early['Close_x'] - arbitrage_early['Close_y']
arbitrage_early.head()
```

```python
#arbitrage_early.loc[arbitrage_early['spread'] >= 0,'rate'] = (arbitrage_early['Close_x'] - arbitrage_early['Close_y'])/arbitrage_early['Close_y']
arbitrage_early.loc[arbitrage_early['spread'] < 0,'rate'] = (arbitrage_early['Close_y'] - arbitrage_early['Close_x'])/arbitrage_early['Close_x']
arbitrage_early.head()
```

```python
df3 = arbitrage_early.loc[arbitrage_early['spread'] < 0]
df3.info()
```

```python
df3.head()
```

#### 3. For each of the three dates, narrow down your trading opportunities even further. To do so, determine the number of times your trades with positive returns exceed the 1% minimum threshold that you need to cover your costs.

```python
df3['profit'] = df3.loc[df3['rate'] >= 0.01,'rate'] 
df3.head()
```

```python
df4 = df3.loc[df3['rate'] >= 0.01]
df4.head()
```
